```python
import chromadb
client = chromadb.Client()
collection = client.create_collection(name="my_collection")
```

```python
collection.add(
    documents=[
        "This document is about New York",
        "This document is about Ibadan"
    ],
    ids = ["id1", "id2"]
)
```

```python
all_docs = collection.get()
all_docs
```

```python
documents = collection.get(ids=["id1"])
documents
```

```python
collection.query(
    query_texts=["Query is about Nigeria"],
    n_results=2
)
```
